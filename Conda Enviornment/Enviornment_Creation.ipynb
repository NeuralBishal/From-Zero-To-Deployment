Creating a Virtual Enviornment is very important before developing a software. There are many ways to create a virtual enviornment. In our course we will use Ananconda

```python
#Create a New enviornment 

conda create -n myenv    #here we using the create command to tell conda we need to create a new enviornment.
                         # -n here represents --name of the enviornment you want to create. we need to add this before our enviornment name
                         # myenv here is our enviornment name(you can provide any name you want. But try to keep the name short and related to your project. it will be easy to remember) 
                         # It will start creating a virtual enviornment. In Between you will be given a choice to continue between Y and N. 
                         # You have to type Y then enter if you want to continue your enviornment creation. Y means Yes and N means No
```

```python
#create a new enviornment with a specific python version

conda create -n myenv python=3.11  # For creating the enviornment in a specific python version we have to type python = (your python version)
                                   # It will give conda instruction to create the enviornment in your specific python version
```

```python
# Activate the enviornment

conda activate myenv    # After enviornment creation is done we have to activate it using the command conda activate. Then your enviornment name.
```

```python
#Create an environment and install packages immediately

conda create -n myenv python=3.11 pandas numpy  # Here we wanted to install pandas and numpy along with the enviornment creation. 
                                                # That's why we typed their names along with the creation command
```

```python
# Deactivate (exit) the current environment

conda deactivate    # This command will deactivate the currently active conda environment and return you to the base environment
```

```python
# List all your environments

conda env list    # This command will display a list of all your conda environmentsList all your environments

xonda info --envs # This command will display a list of all your conda environmentsList all your environments
```

```python
# Clone an existing enviornment

conda create --name newenv --clone oldenv  # This command will clone your old enviornment to your new enviornment with all the dependencies
```

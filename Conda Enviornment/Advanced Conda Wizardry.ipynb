#  How to Compare Two Environments

If you want to compare two different environments, you would need to:

Export each environment to a specification file:

bash
conda env export -n env1 > env1.yml
conda env export -n env2 > env2.yml

Then manually compare the two .yml files, or use a different tool. The search results mention a third-party plugin that can compare two environments, but this is not a built-in conda command .


```python
# Create an environment without installing Python

conda create -n myenv --no-python           # Normally when we dont specify any python enviornment while creating the enviornment it installs a default python version. but if you want to create your enviornment without any python you can use this
```

```python
# See what packages are available in conda-forge

conda search -c conda-forge *                   #be careful, this is a LOT
```

```python
# Cache packages for offline install

conda download package_name        # Did you ever thought you can install something without the internet connection? Well now you can. 
                                   # This command lets you download a package and saves it into cache memory 
```

```python
# Install from local cache (offline)

conda install package_name --offline        # From the cache memory you can install your downloaded package without any internet connection
```

```python
# Find which environment is active	

conda info --envs | grep "*"                            # This command lets you see which enviornments are active. * represents all
```

```python
# Save your environment and project files in one go

conda env export > environment.yml && git add environment.yml           # you can save your enviornment and project files into the github together with this command.
```

```python
List all packages in all environments

for env in $(conda env list | awk '{print $1}'); do echo "=== $env ==="; conda list -n $env; done           # You can see every kinds of packages from every enviornment with this command 
```

```python
# Auto-activate environment on cd

Add this to ~/.zshrc or ~/.bashrc:

cd() { builtin cd "$@"; if [ -f .conda_env ]; then conda activate $(cat .conda_env); fi }           # This command lets your enviornment to be auto activated when you open project folder.
```

```python
# Delete all environments except base

conda env list | grep -v base | awk '{print $1}' | xargs -n1 conda env remove -n        # if you want to delete every enviornment you created except ypur base enviornment you can do that using this command. 
```

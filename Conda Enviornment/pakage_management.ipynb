```python
# Install a Package

conda install pandas. # We can use the conda install command to install any package. Just have to use the package name after this command
```

```python
# Install multiple packages at once

conda install pandas matplotlib numpy   # We can install multiple packages through one single command by specifying the packages names with spaces
```

```python
# Install a specific version

conda install pandas=1.5.3        # We can install a specific version of a particular package just by specifying it using =
```

```python
# Install from the conda-forge channel

conda install -c conda-forge scikit-learn   # We can install packages from conda forge channel using this command. Here -c represents --channel
                                            # The -c flag tells conda which channel to search for the package in. A channel is essentially a repository where conda looks for packages.
```

```python
# Set conda-forge as default (so you don't type -c every time)

conda config --add channels conda-forge     # We can add conda-forge as the default channel for installation using this command
```

```python
# Remove a package

conda remove numpy     # We can remove any existing package from our enviornment using the remove command and then specifying the name of the package we want to remove
```

```python
# Update a package to the latest version

conda update numpy          # We can update any existing package from our enviornment using the update command. 
```

```python
# Update ALL packages in the current environment

conda update --all          # If we want to update all the packages from our enviornment to the latest version at once we can use the update --all command
```

```python
# Search for a package (without installing)

conda search pandas     # We can search for any package without installing it using the command search
```

```python
# See detailed info about a package

conda info pandas          # To see all the information about a package we use the command info. Then name of that particular package
```

```python
# See what's installed in the current environment

conda list          # We can see which packages or dependencies are installed in the current enviornment using this command list
```

```python
# See what's installed in another environment

conda list -n myenv         # If we are in a different enviornment but we want to see what things are installed in another enviornment without switching to that enviornment we can use this command. 
                            # Here we are specifying the name of the enviornment we want to see after the list command
```

```python
# See explicit versions of everything

conda list --explicit       # We can see the explicit versions of everything by specifying --explicit after list.
```

# What conda list --explicit Shows You

When you run this command, it lists every package in your current environment as a full URL pointing to the exact .conda or .tar.bz2 file that was installed for that specific version and build.

It doesn't just give you package names and versions (like numpy=1.24.3). It shows you the complete, specific path to the exact file that conda uses to install that package.

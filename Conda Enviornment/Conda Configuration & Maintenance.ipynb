# Keep your Conda installation healthy and fast.

```python
# Update Conda itself (the tool, not packages)

conda update conda          # This command tells conda to updte itself to the latest version
```

```python
# Update the conda-forge channel's package index

conda update conda-build     # You can update the packages of the conda forge channel by this command
```

```python
# Clean package cache (frees up GBs of space!)

conda clean -a              # More amount of cache can slow down your operations and system. Its better to remove them.
                            # You can use this command to free up huge amount of space
```

```python
# See all Conda settings

conda config --show         # You can see your anaconda settings using this command
```

```python
# Set conda to NOT auto-activate base environment

conda config --set auto_activate_base false     # Sometimes it happens that conda automatically activates the base enviornment To stop this you can use this command
```

```python
# Add a new channel permanently

conda config --add channels newchannel          # If you want to add a channel permanently so that in future you dont have to add it you can use this command
```

```python
# Remove a channel

conda config --remove channels newchannel       # If you want to remove any existing channel you can use this command
```

```python
# See which channel a package came from

conda list | grep package_name                  # Sometimes we need to know which packages came from which channel. This command lets you identify that. 
```

# How It Works (Step by Step)

Step	Command	What It Does
1	    conda list	Lists ALL installed packages in your current environment
2	    | (pipe)	Takes the output from the left command and passes it to the right command
3	    grep package_name	Searches through that output and only shows lines containing "package_name"

# Example

Check if numpy is installed

conda list | grep numpy

# Output:

numpy                     1.26.4          py311h7f2c7f6_0    conda-forge
